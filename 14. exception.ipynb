```python
# ZeroDivisionError

num1 = int(input("enter num one"))
num2 = int(input("enter num two"))

try:
    res = num1/num2

except ZeroDivisionError as err:
    print("can not divide by zero",err)

else:
    print("division of num1 and num2 is :",res)

finally:
    print("division complete")

```
