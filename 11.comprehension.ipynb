```python
listOne = [1,2,3,4,5,6,7,8]
listTwo = [ ]

for i in listOne:
    if i%2 == 0:
        res = i**2
        listTwo.append(res)

    
print(listOne)
print(listTwo)
```

```python
listOne = [1,2,3,4,5,6,7,8]

#  [ expression ,  iteration  , condition (optional)  ]

listTwo = [ i for i in listOne ]
#         [1 , 2 , 3  , 4]

print(listOne)
print(listTwo)
```

```python
#  [ expression ,  iteration  , condition (optional)  ]
#      Third          ,  First    ,    Second

```

```python
#List Comprehension
ListOne = [1,3,5,9,8,7,4]

ListTwo = [i/2 for i in ListOne if i%2==0  ]

print(ListOne)
print(ListTwo)

```

```python
setOne = {1,2,3,4,5,6,7,8,9}

SetTwo = {i**2 for i in setOne}
#         expression    ,   iteration  ,   condition

print(setOne)
print(SetTwo)
```

```python
ListOne = ["mohit","rohit","ram","Sham"]

DictOne = {i:len(i)  for i in ListOne }
#         {key:value}

print(DictOne)
```

```python
listOne = [1,2,3,4,5,6,7,8]

# {1:1,2:4,3,4,5,6,7,8}

DictTwo = {i:i**3 for i in listOne }
#         expression  ,  iteration

print(DictTwo)

```
