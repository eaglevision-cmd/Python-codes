```python
class upi:
    def pay(self):
        print("payment using upi")


class netbanking(upi):
    def pay(self):
        print("payment using netbanking")


class creditcard(netbanking):
    def pay(self):
        print("payment using creditcard")


obj1 = upi()
obj1.pay()

obj2 = netbanking()
obj2.pay()

obj3 = creditcard()
obj3.pay()
```

```python
class calcy:
    def add(self,num1,num2):
        return num1 + num2

    def add(self,num1 , num2 ,num3 , num4):
        return num1 + num2 + num3 + num4

obj1 = calcy()
print(obj1.add(10,20))

print(obj1.add(10,20,30,40))
```
