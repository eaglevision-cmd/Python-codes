```python
listOne = [1,2,3,4,5,6,7,8,9]
listtwo = []

for i in listOne:
    listtwo.append(i*i)

print(listOne)
print(listtwo)
```

```python
listOne = [1,2,3,4,5,6,7,8,9]

listtwo = list(map(lambda i: i*i , listOne))

print(listtwo)
```

```python
listOne = [1,2,3,4,5,6,7,8,9]

listtwo = list(map(lambda i: i+i , listOne))

print(listtwo)
```

```python
listOne = [1,2,3,4,5,6,7,8,9]

even = list(filter(lambda i: i%2==0 , listOne ))

print(even)
```

```python
from functools import reduce 

listOne = [1,2,3,4,5,6,7,8,9]

addition = reduce(lambda num1,num2 : num1 + num2 , listOne)

print(addition)

print(sum(listOne))

```
