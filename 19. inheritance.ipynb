```python
class grandparent:
    def grand_m(self):
        print("this is grand parent class")

class parent(grandparent):
    def parent_m(self):
        print("this is parent class")

class child(parent):
    def child_m(self):
        print("this is child class")

class demo:
    def demo_m(self):
        print("this is demo class")

class temp(child,demo):
    pass

obj1 = temp()
obj1.grand_m()
obj1.parent_m()
obj1.child_m()
obj1.demo_m()
```
