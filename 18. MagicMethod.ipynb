```python
class Temp:
    def __init__(self,num1):
        self.numone = num1

    def __add__(value1 , value2):
        return value1.numone + value2.numone
             #    40           +   20

    def __sub__(value1 , value2):
        return value1.numone - value2.numone
        #    10           +   20

    def __mul__(value1 , value2):
        return value1.numone * value2.numone

    def __truediv__(self, other):
        return self.numone / other.numone

    def __floordiv__(self, other):
        return self.numone // other.numone

    def __mod__(self, other):
        return self.numone % other.numone


        

obj1 = Temp(44)
obj2 = Temp(10)

print( obj1 + obj2)
print( obj1 - obj2)
print(obj1 * obj2)
print(obj1/obj2)
print(obj1//obj2)
print(obj1%obj2)
# print(<__main__.Temp object at 0x00000157F1E09FD0> + <__main__.Temp object at 0x00000157F1C53110>)
```
